`````python
#1. Mostrar en pantalla las opciones
print ('=========================================')
print ('📔 Agenda de Contactos Interactiva 📔')
print ('=========================================')
print (' 1. Añadir un contacto ')
print (' 2. Buscar un contacto ')
print (' 3. Editar un contacto ')
print (' 4. Eliminar un contacto')
print (' 5. Mostrar todos los contactos')
print ('=========================================')

#2. Pedirle al usuario que elija una opción
opcion = input('Elige una opcion (1-5): ')

#3. Estructura de control según la opción elegida
if opcion == '1':
    nombre = input('Ingresa el nombre del contacto:👤 ')
    telefono = input('Ingresa el teléfono del contacto:☎️ ')
    agenda[nombre] = telefono #Diccionario
    print (f"\n✅ Contacto '{nombre}' añadido con éxito")
    
elif opcion == '2':
    nombre = input('Ingresa el nombre del contacto a buscar: 🔍 ')
    if nombre in agenda:
        print(f"\n📞 El teléfono solicitado es: {agenda[nombre]}")
    else:
        print(f"\n❌ Error: El contacto '{nombre}' no se encuentra en la agenda.")
        
elif opcion == '3':
    nombre = input ('Ingresa el nombre del contacto a editar: ')
    if nombre in agenda:
        nuevo_telefono = input (f"Ingresa el nuevo para {nombre}: ")
        agenda[nombre] = nuevo_telefono
        print(f"\n🔄 Teléfono de '{nombre}' actualizado con éxito!")
    else:
        pritn(f"\n❌ Error: No se puede editar porque '{nombre}' no existe en la agenda.")
        
elif opcion == '4':
    nombre = input ('Ingresa el nombre del contacto a eliminar: ')
    if nombre in agenda:
        # El método .pop() elimina la clave del diccionario y devuelve su valor
        agenda.pop(nombre)
        print(f"\n🗑️ Contacto 'nombre' eliminado de la agenda.")
    else:
        print(f"\n❌ Error: El contacto '{nombre}' no existe en la agend.")
 
elif opcion == "5": #Rediseño del punto #5
    if len(agenda) == 0:
        print("\n📭 La agenda está vacía. ¡Añade tu primer contacto con la opción 1!")
    else:
        print("\n=========================================")
        print(f"📋 DIRECTORIO DE CONTACTOS ({len(agenda)} guardados)")
        print("=========================================")
        
        # Transformamos el diccionario a una lista visual limpia sin llaves ni comillas
        contactos_limpios = str(agenda).replace("{", "  • ").replace("}", "").replace(", ", "\n  • ").replace("'", "").replace(":", " 📞 ->")
        print(contactos_limpios)
        
        print("=========================================")

else:
    print(f"\n⚠️ Opción no válida. Por favor, ejecuta de nuevo y elige un número del 1 al 5.")
`````


# 🕵️‍♂️ Análisis de QA: ¿Por qué funciona así?

• Validación con if nombre in agenda:: Antes de buscar, editar o eliminar, verificamos que la clave exista. Si no lo hiciéramos y alguien escribiera un nombre inexistente, Python detendría el programa con un error crítico (KeyError).
• Uso de .pop(nombre): Cumple con la pista de investigar métodos de diccionarios (dir(dict)) para eliminar un registro específico de forma segura.
• Persistencia en Jupyter: Al dejar agenda = {} en una celda separada arriba, puedes correr la segunda celda 3 veces seguidas para añadir 3 contactos distintos, y luego correrla una cuarta vez eligiendo la opción 5 para verlos todos guardados.


Para este proyecto se usó un diccionario (dict), y es la mejor estructura de datos de las tres porque te permite asociar cada nombre directamente con su teléfono mediante pares clave-valor ({"Nombre": "Teléfono"}), haciendo que buscar, editar o borrar sea inmediato.

### ¿Por qué el diccionario es el ganador aquí?

Si comparamos cómo funcionaría tu agenda con cada estructura, la diferencia se ve muy clara:

| Estructura          | ¿Cómo guardaría los datos?                  | ¿Por qué funciona o por qué falla en la agenda?                                                                                                                        |
|---------------------|---------------------------------------------|------------------------------------------------------------------------------------------------------------------------------------------------------------------------|
| Diccionario ({}) ✅ | {"Juan": "3312345678", "Ana": "3398765432"} | Ideal: Es mutable (permite añadir, editar y borrar) y te deja buscar directo por el nombre (agenda["Juan"]) sin saber en qué posición numérica está.                   |
| Lista ([]) ⚠️       | ["Juan", "3312345678", "Ana", "3398765432"] | Poco práctica: Solo guarda elementos sueltos por posición (0, 1, 2...). Para editar el teléfono de "Ana", primero tendrías que adivinar en qué índice quedó su número. |
| Tupla (()) ❌       | ("Juan", "3312345678")                      | No sirve: Es inmutable. Una vez creada, Python prohíbe añadir contactos nuevos, editar teléfonos o eliminar registros.                                                 |


# 🗒 Nota: 3 razones sencillas y atinadas

1. Relación directa (Clave → Valor): El nombre actúa como una etiqueta única (clave) que apunta directo a su teléfono (valor), igual que una agenda física ordenada alfabéticamente.
2. Es mutable: A diferencia de las tuplas, los diccionarios permiten agregar contactos nuevos (agenda[nombre] = telefono), actualizar números existentes y eliminar registros (.pop(nombre)) sobre la marcha.
3. Búsqueda instantánea: Con el operador in (if nombre in agenda:), Python verifica en una sola línea si el contacto existe sin necesidad de recorrer toda la estructura elemento por elemento.

### 🧠 Justificación Técnica: ¿Por qué usar un Diccionario?

Para el almacenamiento de datos de esta agenda se eligió un **Diccionario (`dict`)** por encima de las listas y tuplas debido a tres ventajas clave:

* **Asociación directa (Clave → Valor):** Permite vincular cada nombre (clave única) directamente con su número de teléfono (valor), haciendo que las búsquedas sean inmediatas sin depender de posiciones numéricas.
* **Mutabilidad:** A diferencia de las **tuplas** (que son inmutables y no permitirían agregar ni editar números), el diccionario permite añadir, actualizar y eliminar registros dinámicamente.
* **Eficiencia de consulta:** Frente a las **listas** (donde habría que rastrear índices manualmente para saber qué teléfono pertenece a quién), el diccionario valida la existencia de un contacto en una sola línea mediante el operador `in`.
